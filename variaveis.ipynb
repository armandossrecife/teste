## Variáveis em Python

### O que são Variáveis?
Variáveis são nomes que fazem referência a valores armazenados na memória.

* Em Python, **não é necessário declarar o tipo** da variável antes de usá-la;
* O tipo é definido **automaticamente** conforme o valor atribuído.

### Exemplo de Declaração:
```python
nome = "João"       # str
idade = 30         # int
altura = 1.75      # float
estudante = True   # bool
```

````python
curso = "Ciência da Computação"
````

````python
print(curso)
````

## Boas Práticas na Criação de Variáveis

### Regras e Recomendações:
* **Use nomes claros e descritivos:**
  * `nome_completo`
  * `idade_aluno`
* **Não inicie nomes com números:**
  * `1nome` (Incorreto)
  * `nome1` (Correto)
* **Evite caracteres especiais:**
  * Não use hífen (`-`) ou espaços nos nomes de variáveis.
* **Respeite as palavras reservadas da linguagem:**
  * Não use palavras como `class`, `def`, `if`, `else`, etc.

## Principais Tipos de Dados

| Tipo | Descrição | Exemplo |
| :--- | :--- | :--- |
| `int` | Número inteiro | `42` |
| `float` | Número decimal (ponto flutuante) | `3.14` |
| `str` | Texto (*string*) | `"Python"` |
| `bool` | Valor lógico (*booleano*) | `True` ou `False` |
| `list` | Coleção ordenada e mutável | `[1, 2, 3]` |
| `tuple` | Coleção ordenada e imutável | `(1, 2, 3)` |
| `dict` | Dicionário (pares de chave e valor) | `{"nome": "Ana"}` |
| `set` | Conjunto (coleção sem duplicatas) | `{1, 2, 3}` |


## Verificação do Tipo de Dado

### Como checar o tipo de uma variável em Python?
Utilizamos a função embutida `type()`.

```python
variavel = 100
print(type(variavel))  # Saída: <class 'int'>

variavel_texto = "Olá, Mundo!"
print(type(variavel_texto))  # Saída: <class 'str'>
```



## Operadores Aritméticos

Considerando `x = 10` e `y = 3`:

| Operação | Operador | Exemplo em Python | Resultado |
| :--- | :---: | :--- | :---: |
| Adição | `+` | `x + y` | `13` |
| Subtração | `-` | `x - y` | `7` |
| Multiplicação | `*` | `x * y` | `30` |
| Divisão | `/` | `x / y` | `3.333...` |
| Divisão Inteira | `//` | `x // y` | `3` |
| Resto da Divisão | `%` | `x % y` | `1` |
| Potenciação | `**` | `x ** y` | `1000` |


## Operadores de Comparação

Retornam um valor booleano (`True` ou `False`):

Considerando `a = 5` e `b = 10`:

```python
a == b   # Igual a: False
a != b   # Diferente de: True
a < b    # Menor que: True
a >= b   # Maior ou igual a: False
```


## Operadores Lógicos

Permitem combinar condições lógicas:

| Operador | Descrição | Exemplo |
| :--- | :--- | :--- |
| `and` | Verdadeiro se **ambas** as condições forem verdadeiras | `cond1 and cond2` |
| `or` | Verdadeiro se **pelo menos uma** condição for verdadeira | `cond1 or cond2` |
| `not` | Inverte o estado lógico | `not cond` |

### Exemplo em Código:
```python
idade = 18
tem_cartao = True

pode_entrar = idade >= 18 and tem_cartao  # Resultado: True
```


## Operadores de Atribuição Composta

Simplificam alterações no valor de uma variável existente:

```python
x = 10

x += 5   # Equivalente a x = x + 5  -> Resultado: 15
x *= 2   # Equivalente a x = x * 2  -> Resultado: 30
x -= 3   # Equivalente a x = x - 3  -> Resultado: 27
```


## Conversão de Tipos (Casting)

Podemos converter valores de um tipo para outro usando funções de conversão:

```python
int(3.9)    # Converte para inteiro -> 3 (trunca as casas decimais)
float(10)   # Converte para float   -> 10.0
str(42)     # Converte para string  -> "42"
bool(1)     # Converte para bool    -> True
```


## Tipagem Dinâmica e Mutabilidade

### Tipagem Dinâmica
A mesma variável pode receber valores de tipos diferentes ao longo do programa:

```python
x = 10       # Aqui x é um inteiro (int)
x = "texto"  # Agora x é uma string (str)
```

### Mutabilidade de Tipos em Python
* **Imutáveis:** Strings, tuplas e números (não podem ser alterados após criados).
* **Mutáveis:** Listas e dicionários (podem sofrer alterações internas de conteúdo).
