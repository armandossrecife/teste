## Estruturas de Seleção (if) - Sintaxe Básica

```python
if condição:
    # código se verdadeiro
else:
    # código se falso
```

---

## Exemplo - Estrutura if


````python
idade = 18
if idade >= 18:
    print("Maior de idade")
else:
    print("Menor de idade")
````

### Exemplo com múltiplas condições:

````python
nota = 85
if nota >= 90:
    print("Aprovado com A")
elif nota >= 80:
    print("Aprovado com B")
elif nota >= 70:
    print("Aprovado com C")
else:
    print("Reprovado")
````

## Estruturas de Repetição - for - Sintaxe Básica

```python
for elemento in sequencia:
    # código a executar
```

## Exemplos - Loop for

### Contagem simples:


````python
for i in range(5):
    print(i)
````

### Com passo personalizado:

````python
for i in range(1, 10, 2):
    print(i)
````

### Iteração com índices:


````python
numeros = [10, 20, 30]

for indice, valor in enumerate(numeros):
    print(f"Índice: {indice}, Valor: {valor}")
````

## Slide 5: Estruturas de Repetição - while - Sintaxe Básica

```python
while condição:
    # código enquanto verdadeiro
```


## Exemplo - Loop while

### Loop simples com contador:

````python
contador = 0
while contador < 5:
    print(contador)
    contador += 1
````

### Com break e continue:

````python
i = 0

while i < 10:
    i += 1
    if i == 5:
        continue  # pula o 5
    if i == 8:
        break  # para no 8
    print(i)
````

## Exemplo Integrado - Soma de Números Pares

````python
numeros = [1, 2, 3, 4, 5]
soma = 0

for numero in numeros:
    if numero % 2 == 0:  # verifica se é par
        soma += numero

print(f"Soma dos pares: {soma}")
````

## Slide 8: Exercício Prático 1 - Verificação de Número

````python
numero = int(input("Digite um número: "))

if numero > 0:
    print("Positivo")
elif numero < 0:
    print("Negativo")
else:
    print("Zero")
````

## Exercício Prático 2 - Tabela de Multiplicação

````python
for i in range(1, 11):
    for j in range(1, 11):
        print(f"{i * j:2}", end=" ")
    print()
````

## Exercício Prático 3 - Número Primo

````python
numero = int(input("Digite um número: "))
eh_primo = True

for i in range(2, int(numero / 2) + 1):
    if numero % i == 0:
        eh_primo = False
        break

if eh_primo and numero > 1:
    print("É primo")
else:
    print("Não é primo")
````

## Exercício Prático 4 - Jogo de Adivinhação

````python
import random

numero_aleatorio = random.randint(1, 100)

while True:
    palpite = int(input("Adivinhe (1-100): "))
    if palpite > numero_aleatorio:
        print("Muito alto!")
    elif palpite < numero_aleatorio:
        print("Muito baixo!")
    else:
        print("Acertou!")
        break
````
