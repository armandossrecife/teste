# Entrada e Saída de Dados em Python

## Introdução à Entrada e Saída (E/S)

### O que é Entrada e Saída?
A interação com o usuário é fundamental no desenvolvimento de software.

* **Entrada de Dados (Input):** Mecanismo que permite ao programa receber dados fornecidos pelo usuário ou por fontes externas.
* **Saída de Dados (Output):** Mecanismo que exibe informações, resultados ou mensagens processadas pelo programa na tela.

Em Python, as funções internas essenciais para essas operações são:
* `input()` para entrada de dados.
* `print()` para saída de dados.

## A Função `print()` - Exibindo Informações

### Função `print()`
Sua principal utilidade é imprimir mensagens e valores no console/terminal.

````python
# Exemplo básico
print("Olá, bem-vindo ao curso de Python!")
````

### Exibindo múltiplos valores:
Podemos passar múltiplos argumentos separados por vírgula. Por padrão, eles são separados por um espaço.

````python
nome = "Maria"
idade = 25

print("Nome:", nome, "- Idade:", idade)
````

## Parâmetros Especiais do `print()`: `sep` e `end`

A função `print()` possui parâmetros nomeados que alteram o seu comportamento padrão.

### Parâmetro `sep` (Separador):
Define qual caractere separa os argumentos transmitidos. O valor padrão é um espaço (`" "`).

````python
print("2026", "08", "18", sep="-")

print("python", "programacao", "aula", sep=" @ ")
````

### Parâmetro `end` (Finalizador):
Define o caractere impresso no final da linha. O valor padrão é a quebra de linha (`"\n"`).


````python
print("Carregando", end="...")
print(" Concluído!")
````

## Formatação de Saída com f-strings

A partir do Python 3.6, as **f-strings** (*formatted string literals*) tornaram-se a forma recomendada e mais legível de formatar texto.

### Sintaxe Básica:
Coloca-se a letra `f` ou `F` antes das aspas e insere-se variáveis/expressões entre chaves `{}`.


````python
produto = "Notebook"
preco = 3500.50
quantidade = 2

# Interpolação de variáveis e expressões
print(f"Produto: {produto} | Total: R$ {preco * quantidade}")
# Saída: Produto: Notebook | Total: R$ 7001.0
````

### Formatação Numérica (Casas Decimais):

````python
pi = 3.14159265
print(f"Valor de Pi formatado: {pi:.2f}")
# Saída: Valor de Pi formatado: 3.14
````

## A Função `input()` - Recebendo Dados do Usuário

### Função `input()`
Pausa a execução do programa e aguarda o usuário digitar um valor e pressionar **Enter**.

````python
nome = input("Digite o seu nome: ")
print(f"Olá, {nome}! Seja bem-vindo.")
````

### Ponto Crítico do `input()`:
> **A função `input()` sempre retorna uma STRING (`str`)**, independentemente do que o usuário digitar (números, texto, etc.).

````python
idade = input("Digite sua idade: ")
print(type(idade))  # Saída: <class 'str'>
````

## Slide 6: Conversão de Tipos (Casting) na Entrada de Dados

Para realizar operações matemáticas com valores recebidos pelo `input()`, precisamos realizar a **conversão explícita de tipos**.

### Entrada de Números Inteiros (`int`):



````python
idade = int(input("Digite sua idade: "))
print(f"Ano que vem você terá {idade + 1} anos.")
````

### Entrada de Números Decimais (`float`):



````python
altura = float(input("Digite sua altura em metros (ex: 1.75): "))
print(f"Sua altura é {altura}m.")
````

### Exemplo de Erro Comum sem Conversão:
```python
x = input("Digite um número: ")  # Se o usuário digitar 5 -> x = "5"
y = input("Digite outro número: ") # Se o usuário digitar 3 -> y = "3"

print(x + y)  # Ocorre CONCATENAÇÃO de strings!
# Saída: "53" (em vez de 8)
```

## Exemplo Prático Integrado

### Calculadora Simples de Média Acadêmica

```python
# Entrada de dados com conversão para float
aluno = input("Nome do aluno: ")
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))

# Processamento
media = (nota1 + nota2) / 2

# Saída de dados formatada
print("-" * 30)
print(f"Aluno: {aluno}")
print(f"Média Final: {media:.1f}")

# Validação com estrutura condicional
if media >= 7.0:
    print("Situação: APROVADO")
else:
    print("Situação: RECUPERAÇÃO")
print("-" * 30)
```


### Boas Práticas de Entrada e Saída:
1. **Mensagens Claras no `input()`:** Sempre forneça instruções precisas ao usuário sobre o formato esperado (ex: *"Digite a data no formato DD/MM/AAAA"*).
2. **Tratamento de Conversão:** Certifique-se de converter os valores de entrada para o tipo adequado antes de realizar cálculos.
3. **Legibilidade com f-strings:** Dê preferência às f-strings para construir textos complexos com variáveis.
4. **Organização da Saída:** Utilize argumentos como `sep` e `end` para melhorar a apresentação visual no terminal.
